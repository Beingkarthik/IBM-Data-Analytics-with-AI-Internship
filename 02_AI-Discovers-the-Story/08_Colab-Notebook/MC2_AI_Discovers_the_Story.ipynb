```python
import pandas as pd

# Load the cleaned dataset
df = pd.read_excel("Cleaned_Dataset.xlsx")

# 1. Number of rows
print("Number of rows:", df.shape[0])

# 2. Number of columns
print("Number of columns:", df.shape[1])

# 3. Column names
print("\nColumn names:")
print(df.columns.tolist())

# 4. Data types
print("\nData types:")
print(df.dtypes)

# 5. Summary statistics
print("\nSummary statistics:")
display(df.describe(include="all"))

# 6. Unique values in important categorical columns
print("\nUnique Categories:")
print(df["Category"].unique())

print("\nUnique Regions:")
print(df["Region"].unique())

print("\nUnique Customer Segments:")
print(df["Customer_Segment"].unique())

# 7. First five rows
print("\nFirst five rows:")
display(df.head())
```

```python
print("Customer Segments:")
print(df["Customer_Segment"].unique())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Make a copy of the dataset
monthly_df = df.copy()

# Remove rows where Order_Date is missing
monthly_df = monthly_df.dropna(subset=["Order_Date"])

# Extract month number and month name
monthly_df["Month_Number"] = monthly_df["Order_Date"].dt.month
monthly_df["Month_Name"] = monthly_df["Order_Date"].dt.strftime("%B")

# Calculate total revenue for each month
monthly_revenue = (
    monthly_df.groupby(["Month_Number", "Month_Name"])["Revenue"]
    .sum()
    .reset_index()
    .sort_values("Month_Number")
)

# Display monthly revenue
print("Monthly Revenue:")
display(monthly_revenue)

# Find highest and lowest revenue months
highest_month = monthly_revenue.loc[monthly_revenue["Revenue"].idxmax()]
lowest_month = monthly_revenue.loc[monthly_revenue["Revenue"].idxmin()]

print("\nHighest Revenue Month:")
print(highest_month["Month_Name"], "₹", highest_month["Revenue"])

print("\nLowest Revenue Month:")
print(lowest_month["Month_Name"], "₹", lowest_month["Revenue"])

# Create line chart
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_revenue["Month_Name"],
    monthly_revenue["Revenue"],
    marker="o"
)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
# Calculate total revenue by category
category_revenue = (
    df.groupby("Category")["Revenue"]
    .sum()
    .reset_index()
    .sort_values("Revenue", ascending=False)
)

# Display results
print("Category Revenue:")
display(category_revenue)

# Find highest and lowest revenue categories
highest_category = category_revenue.iloc[0]
lowest_category = category_revenue.iloc[-1]

print("\nHighest Revenue Category:")
print(highest_category["Category"], "₹", highest_category["Revenue"])

print("\nLowest Revenue Category:")
print(lowest_category["Category"], "₹", lowest_category["Revenue"])

# Create bar chart
plt.figure(figsize=(9, 5))

plt.bar(
    category_revenue["Category"],
    category_revenue["Revenue"]
)

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Count unique customers in each customer segment
customer_segments = (
    df.groupby("Customer_Segment")["Customer_ID"]
    .nunique()
    .reset_index(name="Unique_Customers")
    .sort_values("Unique_Customers", ascending=False)
)

# Calculate percentage of customers in each segment
total_customers = customer_segments["Unique_Customers"].sum()

customer_segments["Percentage"] = (
    customer_segments["Unique_Customers"] / total_customers * 100
)

# Display results
print("Customer Segment Analysis:")
display(customer_segments)

# Create bar chart
plt.figure(figsize=(9, 5))

plt.bar(
    customer_segments["Customer_Segment"],
    customer_segments["Unique_Customers"]
)

plt.title("Customer Segments")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Unique Customers")
plt.xticks(rotation=30)
plt.tight_layout()

plt.show()
```

```python
# Calculate total revenue by region
regional_revenue = (
    df.groupby("Region")["Revenue"]
    .sum()
    .reset_index()
    .sort_values("Revenue", ascending=False)
)

# Display results
print("Regional Revenue:")
display(regional_revenue)

# Find highest and lowest revenue regions
highest_region = regional_revenue.iloc[0]
lowest_region = regional_revenue.iloc[-1]

print("\nHighest Revenue Region:")
print(highest_region["Region"], "₹", highest_region["Revenue"])

print("\nLowest Revenue Region:")
print(lowest_region["Region"], "₹", lowest_region["Revenue"])

# Create horizontal bar chart
plt.figure(figsize=(9, 5))

plt.barh(
    regional_revenue["Region"],
    regional_revenue["Revenue"]
)

plt.title("Revenue by Region")
plt.xlabel("Revenue (₹)")
plt.ylabel("Region")

# Display highest revenue region at the top
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
# Calculate total profit by product
product_profit = (
    df.groupby("Product")["Profit"]
    .sum()
    .reset_index()
    .sort_values("Profit", ascending=False)
)

# Select top 10 products
top_10_products = product_profit.head(10)

# Display results
print("Top 10 Products by Profit:")
display(top_10_products)

# Create horizontal bar chart
plt.figure(figsize=(10, 6))

plt.barh(
    top_10_products["Product"],
    top_10_products["Profit"]
)

plt.title("Top 10 Products by Profit")
plt.xlabel("Total Profit (₹)")
plt.ylabel("Product")

# Highest profit product at the top
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```
